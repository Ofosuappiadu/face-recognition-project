## 1. The idea behind linear regression

Linear regression models the relationship between an input and a numerical target. In simple linear regression, the model can be written as:

\[
\hat{y} = b_0 + b_1 x
\]

- \(x\): input feature, such as study hours
- \(\hat{y}\): predicted target, such as exam score
- \(b_0\): intercept
- \(b_1\): slope

A model's predictions are estimates; they are not guaranteed to be correct.

```python
# Import the libraries used in this notebook
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

print("Libraries imported successfully.")
```

## 2. Create and inspect a small dataset

The data below is synthetic. Each row pairs study hours with an example exam score.

```python
data = {
    "study_hours": [1, 2, 3, 4, 5, 6, 7, 8],
    "exam_score":  [42, 48, 55, 59, 65, 71, 75, 83]
}

df = pd.DataFrame(data)
display(df)

print("\nDataset shape (rows, columns):", df.shape)
print("\nSummary statistics:")
display(df.describe())
```

## 3. Visualise the relationship

A scatter plot helps us see how two numerical variables relate to one another.

```python
plt.figure(figsize=(7, 4))
plt.scatter(df["study_hours"], df["exam_score"])
plt.xlabel("Study hours")
plt.ylabel("Exam score")
plt.title("Study Hours vs Exam Score (Synthetic Data)")
plt.grid(True, alpha=0.3)
plt.show()
```

**Interpretation task:** Does the plotted relationship appear generally positive, negative, or absent? Does that alone prove that studying more causes higher scores? Why or why not?

## 4. Prepare the input (`X`) and target (`y`)

In scikit-learn, `X` is normally two-dimensional (a table of features), while `y` is one-dimensional (the target values).

```python
X = df[["study_hours"]]
y = df["exam_score"]

print("X shape:", X.shape)
print("y shape:", y.shape)

display(X.head())
display(y.head())
```

## 5. Train a simple linear regression model

For this introductory exercise, we fit the model on the small dataset and then evaluate its fit on the same data. **This is for learning the mechanics only**; it does not estimate how well the model generalises to new data. Later, practise a train/test split.

```python
model = LinearRegression()
model.fit(X, y)

print("Intercept:", round(model.intercept_, 3))
print("Slope:", round(model.coef_[0], 3))
```

### Interpret the coefficients

- The **slope** estimates how much the predicted exam score changes for each additional study hour in this fitted line.
- The **intercept** is the model's predicted score at zero study hours. Since zero hours is outside this dataset's observed range, interpret it cautiously.

These are associations in synthetic data, not proof of cause and effect.

## 6. Make predictions

```python
predictions = model.predict(X)

results = df.copy()
results["predicted_score"] = predictions.round(2)
results["error"] = (results["exam_score"] - results["predicted_score"]).round(2)

display(results)
```

```python
# Predict a score for 4.5 study hours
new_data = pd.DataFrame({"study_hours": [4.5]})
predicted_score = model.predict(new_data)[0]

print(f"Predicted score for 4.5 study hours: {predicted_score:.2f}")
```

## 7. Evaluate using Mean Squared Error (MSE)

MSE is the average of squared differences between actual and predicted values:

\[
\mathrm{MSE} = \frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2
\]

A smaller MSE means smaller squared errors on the data being evaluated. MSE is expressed in squared target units (here, score-points squared).

```python
mse = mean_squared_error(y, predictions)
print(f"Training-data MSE (same data used for fitting): {mse:.3f}")
```

## 8. Plot the fitted regression line

```python
plt.figure(figsize=(7, 4))
plt.scatter(df["study_hours"], y, label="Actual scores")
plt.plot(df["study_hours"], predictions, label="Fitted regression line")
plt.xlabel("Study hours")
plt.ylabel("Exam score")
plt.title("Simple Linear Regression")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
